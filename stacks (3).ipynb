# 2. stack  자료구조
- ADT 스택 구현
-  객체지향 프로그래밍 기법 (OOP)

```python
%%writefile ArrayStack.py

#======================================================================
# 스택 클래스 정의
#======================================================================

class ArrayStack:
    # 멤버 속성 정의 - 생성자
    def __init__(self, capacity):
        self.capacity = capacity #스택의 용량을 지정
        self.array = [None] * capacity # None으로 채움
        self.top = -1 # 스택 상단의 인덱스를 공백상태 -1로 초기화함

    # 멤버 메서드(클래스 내부 함수) 정의
    def is_empty(self):
        """스택이 비어있는지 확인하는 메서드"""
        # top이 -1이면 데이터가 하나도 없다는 의미이므로 True, 아니면 False를 반환
        return self.top == -1

    def is_full(self):
        """스택이 가득 찼는지 확인하는 메서드"""
        # top이 (용량 - 1)에 도달했으면 마지막 인덱스까지 찬 상태이므로 True, 아니면 False를 반환
        return self.top == self.capacity - 1

    def push(self, item):  # 스택에 데이터 삽입
        if not self.is_full():
            self.top += 1
            self.array[self.top] = item
        else:
            raise OverflowError("스택 오버플로우")  # 예외 처리

    def pop(self):
        if not self.is_empty():
            item = self.array[self.top]
            self.top -= 1
            print(f"POP : {item} -> stack = {self.array[:self.top + 1]}")
            return item
        else:
            raise IndexError("스택 언더플로우")

    def peek(self):
        if not self.is_empty():
            return self.array[self.top]
        else:
            raise IndexError("스택 언더플로우")

    def size(self):
        return self.top + 1

#======================================================================
# 스택 클래스 사용
#======================================================================

if __name__ == "__main__":
    # 스택 생성
    st = ArrayStack(5)

    # 스택에 1 ~ 5 를 스택에 저장하기
    for i in range(1, 6, 1):
        st.push(i)

    # 스택의 마지막 요소 삭제한 후 확인, 크기 (peef -> peek 수정)
    print(st.pop(), st.peek(), st.size())
```

# 3. 스택 자료 구조 응용
- 과호검사 문제

```python
import importlib
import ArrayStack as st # 스택 자료구조 임포트
importlib.reload(st)     # 수정된 ArrayStack.py 모듈을 강제로 다시 로드합니다.

# 괄호 검사 알고리즘 구현
def check_brackets(expr):
    stack = st.ArrayStack(len(expr)) # 스택생성

    # 괄호 매칭
    pairs = {')' : '(', '}' : '{' , ']' : '[' }
    # 닫는 괄호를 넣으면 그에 맞는 '짝꿍 여는 괄호'를 바로 찾아주는 사전(딕셔너리)

    for char in expr:
        if char in "({[":
            stack.push(char)

        elif char in ")}]":  # ")})"에서 대괄호(']')가 들어가도록 수정했습니다.
            if stack.is_empty():
                return False # 여는괄호가 부족한경우

            top_char = stack.pop()

            if pairs[char] != top_char:
              #방금 읽은 닫는 괄호(char)의 짝꿍과, 스택에서 꺼낸 가장 최근의 여는 괄호(top_char)가 일치하지 않는가?"를 검사하는 조건문
                return False # 괄호가 일치하지 않는 경우

    return stack.is_empty() # 순회가 끝난 후 스택이 비어있어야함


if __name__ == "__main__":
    test_cases = [
        "{[A+(B*C)] -D}", "((A + B)" , "({[A+B]})", "A+B)"
    ]

    # 수정 5: if __name__ == "__main__": 블록 안쪽으로 들여쓰기 맞춤
    for expr in test_cases:
        print(f"수식 : {expr} -> 올바른 괄호 여부 : {check_brackets(expr)}")
```
