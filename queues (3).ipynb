# 1. (선형) 큐
- 선형 자료구조
- 배열로 구현
- 삽입 연산 수행시 rear을 증가시켜 처리
  - rear = rear + 1
- 삭제 연산 수행시 front을 증가시켜 처리
  - front = front + 1

```python
#==============================
# 선형큐의 ADT
#==============================
class ArrayLinearQueue:
  def __init__(self,capacity):
    self.capacity = capacity
    self.array = [None]*capacity
    self.rear = -1
    self.front = -1

  def isEmpty(self):
    # 큐가 공백인지 확인
    return self.rear == self.front

  def isFull(self):
    #큐가 포화 상태인지 확인
    return self.rear == self.capacity-1

  def enqueue(self, item):
    # 큐의 맨 뒤(rear)에 요소 추가
    if self.isFull():
      print("Overflow 발생")
      return
    self.rear += 1
    self.array[self.rear] = item

  def dequeue(self):
    # 큐의 맨 앞(front) 요소 반환
    if self.isEmpty():
      print("Underflow 발생")
      return None
    self.front += 1
    item = self.array[self.front]
    self.array[self.front] = None # 삭제 후 빈칸 처리 (메모리 참조 해제)
    return item

  def peek(self):
    # 큐의 맨 앞의 요소 조회
    if self.isEmpty():
      print("Underflow 발생")
      return None
    return self.array[self.front + 1]

  def size(self):
    # 현재 큐에 저장된 요소의 개수
    return self.rear - self.front

  def display(self):
    # 큐의 현재 상태 출력
    print(f"선형 큐 (front={self.front}, rear = {self.rear}: queue = {self.array}, size={self.size()})")

#=======================================
# 선형큐의 테스트
#======================================
if __name__ == "__main__":
  q = ArrayLinearQueue(capacity = 3)

  # 1.초기 상태 검증
  assert q.isEmpty()
  assert q.size() == 0

  # 2. 요소 삽입
  q.enqueue('A')
  q.display()
  q.enqueue('B')
  q.display()
  q.enqueue('C')
  q.display()

  assert q.isFull()
  assert q.peek() == 'A'

  # 3. 요소 삭제
  print(f"Dequeued: {q.dequeue()}")
  q.display()

  # 4. 선형큐의 문제점 확인- FalseOverflaw
  q.enqueue('D') # 앞쪽 인덱스가 비어있지만(None), rear가 2 = capacity - 1 즉, 끝에 도달해서 더 이상 eqnqueue가 불가

```

# 2. 원형큐(Circular Queue)
- 선형구조(배열)를 원형 구조로 변환
  - 배열은 인덱스가 0부터 capacity -1 까지 직선 형태로 구성
    - rear = rear + 1, front = front +1 수행하면 capacity에 도달하면 범위를 벗어나는 인덱스오류(Index out of bounds) 발생
  - 모듈러연산자(%capacity)을 적용하여 인덱스 결과가 0부터 capacity -1 사이의 범위내로 자동 제한
- 인덱스 연산 공식
    - 다음인덱스 = (현재 인덱스 + 1) % capacity
    - 시계방향으로 이동하여 front, rear 연산
    - 예제: 인덱스가 0, 1, 2,3, 4까지 존재
        - rear = 3일 때, 다음 rear = (3 + 1) % 5 = 4 % 5 = 4
        - rear = 4일 때, 다음 rear = (4 + 1) % 5 = 5 % 5 = 0


```python
#==============================
# 원형큐의 ADT
#==============================
class ArrayQueue:
  def __init__(self,capacity):
    self.capacity = capacity
    self.array = [None]*capacity
    self.rear = 0
    self.front = 0

  def isEmpty(self):
    # 큐가 공백인지 확인
    return self.rear == self.front

  def isFull(self):
    #큐가 포화 상태인지 확인(rear가 배열의 끝에 도달했을 때 인덱스가 front일치)
    return self.front == self.rear + 1 % self.capacity

  def enqueue(self, item):
    # 큐의 맨 뒤(rear)에 요소 추가
    if self.isFull():
      print("Overflow 발생")
      return
    self.rear = (self.rear + 1) % self.capacity
    self.array[self.rear] = item

  def dequeue(self):
    # 큐의 맨 앞(front) 요소 반환
    if self.isEmpty():
      print("Underflow 발생")
      return None
    self.front = (self.front + 1) % self.capacity
    item = self.array[self.front]
    self.array[self.front] = None # 삭제 후 빈칸 처리 (메모리 참조 해제)
    return item

  def peek(self):
    # 큐의 맨 앞의 요소 조회
    if self.isEmpty():
      print("Underflow 발생")
      return None
    return self.array[(self.front + 1)% self.capacity]

  def size(self):
    # 현재 큐에 저장된 요소의 개수
    return (self.rear - self.front + self.capacity) % self.capacity

  def display(self):
    # 큐의 현재 상태 출력
    print(f"선형 큐 (front={self.front}, rear = {self.rear}: queue = {self.array}, size={self.size()})")
    out = []
    for i in range(self.size()):
      idx = (self.front + 1 + i) % self.capacity
      out.append(self.array[idx])
    print(out)
#=======================================
# 원형큐의 테스트
#======================================
if __name__ == "__main__":
  q = ArrayQueue(capacity = 3)

  for item in ['A', 'B', 'C', 'D']:
        q.enqueue(item)
  q.display()
```

# 4. 원형 덱
  - 원형 큐를 상속 받아서 클래스 설계
  - 고성능 자료구조 설계

```python
# 1. 부모 클래스  : 원형 큐
class ArrayQueue:
  def __init__(self,capacity):
    self.capacity = capacity
    self.array = [None]*capacity
    self.rear = 0
    self.front = 0

  def isEmpty(self):
    # 큐가 공백인지 확인
    return self.rear == self.front

  def isFull(self):
    #큐가 포화 상태인지 확인(rear가 배열의 끝에 도달했을 때 인덱스가 front일치)
    return self.front == (self.rear + 1) % self.capacity

  def enqueue(self, item):
    # 큐의 맨 뒤(rear)에 요소 추가
    if self.isFull():
      print("Overflow 발생")
      return
    self.rear = (self.rear + 1) % self.capacity
    self.array[self.rear] = item

  def dequeue(self):
    # 큐의 맨 앞(front) 요소 반환
    if self.isEmpty():
      print("Underflow 발생")
      return None
    self.front = (self.front + 1) % self.capacity
    item = self.array[self.front]
    self.array[self.front] = None # 삭제 후 빈칸 처리 (메모리 참조 해제)
    return item

  def peek(self):
    # 큐의 맨 앞의 요소 조회
    if self.isEmpty():
      print("Underflow 발생")
      return None
    return self.array[(self.front + 1)% self.capacity]

  def size(self):
    # 현재 큐에 저장된 요소의 개수
    return (self.rear - self.front + self.capacity) % self.capacity

  def display(self):
    # 큐의 현재 상태 출력
    print(f"선형 큐 (front={self.front}, rear = {self.rear}: queue = {self.array}, size={self.size()})")
    out = []
    for i in range(self.size()):
      idx = (self.front + 1 + i) % self.capacity
      out.append(self.array[idx])
    print(out)

# 2.  자식 클래스 : 덱 - 원형큐
class CircularDeque(ArrayQueue): # (부모 클래스 작성) 부모의 모든 재산을 자식 클래스로 만들겠다 , 자식 클래스 안에 부모 클래스
  def __init__(self,capacity): # CircularDeque의 생성자임
    super().__init__(capacity) # 부모 클래스의 생성자를 호출

  # [후단연산] 부모 클래스 메소드 호출
  def addRear(self,item):
    self.enqueue(item)

  def deleteFront(self):
    self.dequeue()

  def getFront(self):
    return self.peek()

  # 전단/ 후단 연산 및 반시계 방향 회전 연산 - 새로 추가 구현
  def addFront(self,item):
    if not self.isFull():
      self.array[self.front] = item # 포화 상태가 아니면  front에 요소를 삽입
      self.front = (self.front - 1 +self.capacity) % self.capacity #fornt를 반시계 방향을 회전
    else :
      print("Overflow 발생")
      # raise OverflowError("넘칩니다") > 예외 처리

  def deleteRear(self):
    if not self.isEmpty():#큐가 비어있지 않은지 확인합니다
      item = self.array[self.rear] # 삭제될 반환 요소 - 현재 rear(맨 뒤) 위치에 있는 데이터를 item이라는 변수에 임시로 저장
      self.array[self.rear] = None # 삭제 후 빈칸처리 (메모리 참조 해제) － 데이터를 꺼낸 원래 자리는 비워둡니다(None).
      self.rear = (self.rear - 1 + self.capacity) % self.capacity #맨 뒤(rear)를 가리키는 위치를 한 칸 앞으로(왼쪽으로) 당깁니다
      return item
    else:
      print("언더플로우요")

  def getRear(self):
      if not self.isEmpty():
        return self.array[self.rear] # 공백이 아니면 rear 요소를 반환
      else:
        print("언더플로우요")

#=======================================
# 원형덱 테스트
#======================================
# 1. 크기가 8인 덱 객체 생성
dq = CircularDeque(capacity=8)
dq.display()

# 2. 후단 삽입(addRear)  &  전단 삽입 (addFront)
dq.addRear("A") #rear :  시계 방향으로 이동
dq.display()
dq.addFront("B") #front : 반시계 방향이동
dq.display()

# 3.  후단 삭제 (deleteRear)
dq.deleteRear()
dq.display()

# 4.  전단 삭제 (deleteFront)
dq.deleteFront()
dq.display()
```
