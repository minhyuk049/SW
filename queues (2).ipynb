# 1.(선형) 큐
- 선형 자료구조
- 배열로 구현
- 삽입 연산 수행시 rear을 증가시켜 처리
  - rear = rear + 1
- 삭제 연산 수행시 front룰 증가시켜 처리
  - front = front + 1

```python
# ==========================================
# 선형 큐의 ADT - 추상자료형
# ==========================================
class ArrayLinearQueue:
    def __init__(self,capacity):
      self.capacity = capacity
      self.array = [None]*capacity
      self.rear = -1
      self.front = -1

    def isEmpty(self):
      # 큐가 공백인지 확인
      return self.rear == self.front

    def isFull(self):
      # 큐가 포화상태인지 확인
      return self.rear == self.capacity - 1

    def enqueue(self,item):
      # 큐의 맨뒤(rear)에 요소 추가
      if self. isFull():
        print("Overflow 발생")
        return
      self.rear +=  1
      self.array[self.rear] = item

    def dequeue(self):
      # 큐의 맨앞(front)요소 반환
      if self.isEmpty():
        print("underflow 발생")
        return None
      self.front += 1
      item = self.array[self.front]
      self.array[self.front] = None # 삭제 후 빈칸처리(메모리 참조 해제)
      return item

    def peek(self):
      # 큐의 맨앞의 요소 조회
      if self.isEmpty():
        print("underflow 발생")
        return None
      return self.array[self.front+1]

    def size(self):
      # 현재 큐에 저장된 요소의 개수
      return self.rear - self.front

    def display(self):
      # 큐의 현재 상태 출력
      print(f"선형큐(front = {self.front}, rear = {self.rear} : queue = {self.array} , size = {self.size()})")

# ==========================================
# 선형 큐의 테스트
# ==========================================
if __name__ =="__main__": # capacity는 큐가 저장할 수 있는 '최대 데이터 개수(용량)'
  q = ArrayLinearQueue(capacity = 3) #none ,none, none 이래 됨

  # 1, 초기 상태 검증
  assert q.isEmpty()
  assert q.size() == 0

  # 2. 요소 삽입
  q.enqueue('A')
  q.display()
  q.enqueue('B')
  q.display()
  q.enqueue('C')
  q.display()

  assert q.isFull()
  assert q.peek() == 'A'

  # 3. 요소 삭제
  print(f"Dequeue : {q.dequeue()}")
  q.display()

  # 4. 선형큐의 문제점 확인  - FlaseOverflow
  q.enqueue('D') # 앞쪽 인덱스가 비어있지만(None) rear가 2 = capacity -1 즉, 끝에 도달해서  더 이상 enqueue가 불가
```

2.  원형 큐 (Circular Queue)
- 선형구조(배열)을 원형 구조로 변환
  - 배열은 인덱스가 0부터 capcacity -1 까지 직성 형태로 구성
    - rear = rear +1, front = front +1 수행하면 capcacity에 도달하면 범위를 벗어나는 인덱스오류 (index out of bounds) 발생
  - 모듈러 연산자 (%capacity)를 적용하여 인데스 결과가 0부터 capacity -1 사이의 범위 내로 자동 제한
- 인덱스 연산 공식
  - 다음 인텍스 = (현재 인덱스 + 1) % capacity //나머지 구하기
  - 시계방향으로 이동하여 front, rear 연산
  - 예제 : 인덱스가 0,1,2,3,4 까지 존재
    - rear = 3 일 떄, 다음 rear = (3 + 1) % 5 = 4 % 5 -> 4
    - rear = 4 일 떄, 다음 rear = (4 + 1) % 5 = 5 % 5 -> 0

```python
# ==========================================
# 원형 큐의 ADT - 추상자료형
# ==========================================

class ArrayQueue:
    def __init__(self,capacity):
      self.capacity = capacity
      self.array = [None]*capacity
      self.rear = 0
      self.front = 0


    def isEmpty(self):
      # 큐가 공백인지 확인
      return self.front == self.rear

    def isFull(self):
      # 큐가 포화상태인지 확인 (rear가 배열의 끝에 도달했을때의 인덱스가 front와 일치)
      return self.front == (self.rear + 1) % self.capacity

    def enqueue(self,item):
      # 큐의 맨뒤(rear)에 요소 추가
      if self. isFull():
        print("Overflow 발생")
        return
      self.rear = (self.rear + 1) % self.capacity
      self.array[self.rear] = item

    def dequeue(self):
      # 큐의 맨앞(front)요소 반환
      if self.isEmpty():
        print("underflow 발생")
        return None
      self.front = (self.front + 1) % self.capacity
      item = self.array[self.front]
      self.array[self.front] = None # 삭제 후 빈칸처리(메모리 참조 해제)
      return item

    def peek(self):
      # 큐의 맨앞의 요소 조회
      if self.isEmpty():
        print("underflow 발생")
        return None
      return self.array[(self.front+1) % self.capacity]

    def size(self):
      # 현재 큐에 저장된 요소의 개수
      return (self.rear - self.front + self.capacity) % self.capacity

    def display(self):
      # 큐의 현재 상태 출력
      print(f"선형큐(front = {self.front}, rear = {self.rear} : queue = {self.array} , size = {self.size()})")
      out = []
      for i in range(self.size()):
        i = (self.front + 1 + i) % self.capacity
        out.append(self.array[i])
      print(out)


# ==========================================
# 원형 큐의 테스트
# ==========================================
if __name__ =="__main__": # capacity는 큐가 저장할 수 있는 '최대 데이터 개수(용량)'
  q = ArrayQueue(capacity = 3) #none ,none, none 이래 됨

  for item in ['A','B','C','D']:
      q.enqueue(item)
  q.display()
```
