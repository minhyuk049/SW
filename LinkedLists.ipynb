# 1. 단순 연결 구조로 단순연결 리스트 구현
  - 일단 단순 연결 리스트의 구조
    - None -> Node A -> [Node] -> Node B -> None (노드 A가 머리노드, 노드 B가 꼬리노드)
      - Node A  & Node B 를 처리 ( 추가 삭제 연산)을 할 때는 예외검사가 항상 필요 : if문 처리

```python
# ============================================================
# 단순 연결 구조의 노드 클래스
# ============================================================

class Node:
    def __init__(self, elem, link=None):
        self.data = elem                  # 전달받은 데이터(elem)를 객체의 data 멤버 변수에 저장
        self.link = link                  # 전달받은 링크(기본값 None)를 객체의 link 멤버 변수에 저장

    # 2. append 연산: 현재 노드(self) 뒤에 새로운 노드(node)를 중간에 낑궈넣기
    def append(self, node):
        if node is not None:              # [예외 검사 1] 삽입하려는 노드가 비어있지 않은지(None이 아닌지) 안전하게 확인
            node.link = self.link         # 1단계: 새 노드가 기존의 다음 노드를 먼저 안전하게 붙잡도록 연결
            self.link = node              # 2단계: 현재 노드의 링크가 새 노드를 가리키도록 변경

    # 3. popNext 연산: 현재 노드(self) 바로 뒤에 있는 다음 노드를 제거(추출)
    def popNext(self):
        next = self.link                  # 제거할 대상이 되는 바로 다음 노드를 임시 변수 next에 저장
        if next is not None:              # [예외 검사 2] 제거할 다음 노드가 실제로 존재하는지(None이 아닌지) 확인
            self.link = next.link         # 현재 노드의 링크를 '다다음 노드'(next.link)로 직접 연결하여 중간 노드를 건너뜀
        return next.data                  # 제거된 노드에 담겨 있던 데이터를 반환

```

```python
class LinkedList:
  def __init__ (self):
    self.head =  None

  def isEmpty(self):
    return self.head is None

  def isFull(self):
    return False

  def getNode(self, pos):
    if pos < 0 :
      return None
    ptr = self.head

    for _ in range (pos):
      if ptr is None:
        return None
      ptr = ptr.link

    return ptr

  def getEntry (self, pos):
    node =  self.getNode(pos)
    if node is None:
      return None
    return node.data

  def insert(self, pos, e) :
      node = Node(e, None)              # 1. 삽입할 새로운 노드를 만듦
      before = self.getNode(pos-1)      # 2. 삽입할 위치 이전 노드(pos-1) 탐색

      if before is None :               # 3-A. 맨 앞에 삽입하는 경우 (예외 처리)
          node.link = self.head         # 새 노드가 기존의 첫 번째 노드(head)를 가리키게 함
          self.head = node              # 헤드 포인터(self.head)를 새 노드로 변경

      else :                            # 3-B. 중간이나 끝에 삽입하는 경우
          before.append(node)           # before 노드 바로 뒤에 새 노드를 추가

  def delete(self, pos) :
    before = self.getNode(pos-1)          # 1. 삭제할 위치 이전 노드(pos-1) 탐색

    if before is None :                   # 2-A. 맨 앞(머리노드, pos=0)을 삭제하는 경우
        before = self.head                # 삭제되어 사라질 현재 머리노드를 임시 저장 (나중에 반환하기 위함)
        if self.head is not None :        # 리스트가 비어있지 않다면
            self.head = self.head.link    # 머리노드(head)를 한 칸 뒤(다음 노드)로 변경
        return before                     # 삭제된(꺼낸) 옛날 머리노드를 반환

    else:
      return before.popNext()
    # 2-B. 중간이나 끝 노드를 삭제하는 경우: before의 다음 노드 삭제 및 반환

  def size( self ) :
    ptr = self.head       # 1. 탐색용 포인터(ptr)를 맨 앞(head)에 놓음
    count = 0;            # 2. 개수를 셀 변수를 0으로 초기화
    while ptr is not None : # 3. ptr이 빈 공간(None)이 아닐 동안 계속 반복
        ptr = ptr.link    # 다음 노드로 화살표를 따라 이동
        count += 1        # 노드를 하나 지날 때마다 count를 1씩 증가
    return count          # 반복이 끝나면 총 개수 반환

  def display(self, msg='단순 연결 리스트:'):
    """연결 리스트의 전체 요소를 형식에 맞게 출력"""
    print(msg, end=' ')
    ptr = self.head       # 머리노드부터 시작
    while ptr is not None:
        print(ptr.data, end=' -> ')  # 각 노드의 데이터(data)를 출력하고 화살표 표시
        ptr = ptr.link            # 다음 노드로 이동
    print('None')                 # 리스트의 끝을 알리는 None 출력

# ============================================================
# 테스트 코드
# ============================================================

if __name__ == '__main__':
  sl = LinkedList()

  sl.display("[초기 리스트]")

  sl.insert(0, 10)
  sl.display()

  sl.insert(1, 20)
  sl.display()

  sl.insert(1, 15) # 리스트 자료구조는 중간에 삽입할 경우 나머지 노드들이 뒤로 이동
  sl.display()

  sl.insert(sl.size(), 25) # 25 넣기 전 노드의 총 개수는 3이지만 pos = 3
  sl.display()

  print(sl.getEntry(0))
  print(sl.getEntry(1))

  delNode = sl.delete(2)
  print(f"중간노드 삭제 후 : {delNode}")
  sl.display()

#===========================
# 문제 : 출제 예시
#===========================
# 아래의 코드를 구현

def replace (self,pos, elem):
  #pos 위치의 노드 데이터를 elem으로 변경 하는 연산이다
  # 여기에 코드를 작성하시오
  #단, 리스트의 기본 연산을 활용

  node = self.getNode(pos)
  if node is not None:
    node.data = elem

def replace(self, pos, elem):
    # pos 위치의 기존 노드를 삭제하고, 동일 위치에 새 데이터를 삽입
    self.delete(pos)
    self.insert(pos, elem)
```
