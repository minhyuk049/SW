# 2. Stack 자료구조
- ADT 스택 구현 : 배열 이용
- 객체지향프로그래밍 기법 (OOP)

```python
%%writefile ArrayStack.py
#=======================================================
# 스택 클래스 정의 (ArrayStack.py)
#=======================================================
class ArrayStack:
    def __init__(self, capacity):
        self.capacity = capacity
        self.array = [None] * capacity
        self.top = -1

    def isEmpty(self):
        return self.top == -1

    def isFull(self):
        return self.top == self.capacity - 1

    def push(self, item):
        if not self.isFull():
            self.top += 1
            self.array[self.top] = item
        else:
            raise OverflowError("Stack Overflow!!!")

    def pop(self):
        if not self.isEmpty():
            item = self.array[self.top]
            self.top -= 1
            return item
        else:
            raise IndexError("Stack Underflow!!!")

    def peek(self):
        if not self.isEmpty():
            return self.array[self.top]
        else:
            raise IndexError("Stack Underflow!!!")

    def size(self):
        return self.top + 1
```

# 3. 스택 자료구조 응용
- 괄호검사 문제
  - 문제 설명:문자열 형태의 수학 수식을 입력받아 괄호 (), {}, []의 짝이 올바르게 맞는지 검사하는 프로그램을 작성하시오.
  - 조건:
    - 조건 1: 여는 괄호 (, {, [를 만나면 스택에 삽입(push)합니다.
    - 조건 2: 닫는 괄호 ), }, ]를 만나면 스택에서 최상단 요소를 꺼내(pop) 짝이 맞는지 비교합니다.
    - 조건 3: 수식 순회가 끝난 후 스택이 비어 있어야 올바른 괄호입니다. 스택이 비어 있지 않거나, 중간에 짝이 맞지 않으면 오류로 판단합니다.
  - 입출력 예시:
    - 입력: "{[A + (B * C)] - D}" $\rightarrow$ 출력: True (올바른 괄호)
    - 입력: "((A + B)" $\rightarrow$ 출력: False (괄호 불일치)
- Colab 환경에서 사용자 정의 모듈 파일을 사용해서 구현
  - import ArrayStack as st


```python
import importlib
import ArrayStack as st

# 메모리에 저장된 모듈 강제 재로딩
importlib.reload(st)

# 괄호검사 알고리즘 구현
def check_brackets(expr):
    stack = st.ArrayStack(len(expr))
    pairs = {')': '(', '}': '{', ']': '['}

    for char in expr:
        if char in "({[":
            stack.push(char)
        elif char in ")}]" :
            if stack.isEmpty():
                return False

            if pairs[char] != stack.pop():
                return False

    return stack.isEmpty()

# 테스트 실행
if __name__ == "__main__":
    test_cases = [
        "{[A + (B * C)] - D}", # True
        "((A + B)",           # False
        "({[A + B]})",        # True
        "A + B)"              # False
    ]

    for expr in test_cases:
        result = check_brackets(expr)
        print(f"수식: {expr:<20} -> 올바른 괄호 여부: {result}")
```

# 4. 시스템 호출 스택을 이용한 응용문제 (재귀적(순환) 프로그래밍)
  - 재귀적인 문제
    - n!
    - 하노이 탑
    - 전체 파일 용량 구하기

  - 재귀 구조(Recursion) vs 반복 구조(Iteration) 비교
    - **재귀 구조 (Recursion)**: 하나의 함수가 자기 자신을 다시 호출하여 문제를 더 작은 단위의 서브 문제(Sub-problem)로 분할해 해결하는 구조
    - **반복 구조 (Iteration)**: `for`나 `while` 루프 문을 사용하여 지정한 조건이 참(True)인 동안 명시적으로 문장을 반복 실행하는 구조

---

   
| 비교 항목 | 재귀 구조 (Recursion) | 반복 구조 (Iteration) |
| :--- | :--- | :--- |
| **제어 방식** | 자기 자신 함수 호출 (Recursive Call) | 반복문 (`for`, `while`) |
| **종료 조건** | 기저 조건 (Base Case) 미작성 시 **무한 재귀** | 반복 조건 (Loop Condition) 미달성 시 **무한 루프** |
| **메모리 사용** | **콜 스택(Call Stack)** 사용 (스택 프레임 누적) | 추가 메모리 거의 없음 (지정된 변수만 사용) |
| **오버헤드** | 함수 호출 오버헤드 발생 (메모리, 시간) | 오버헤드가 적고 실행 속도가 빠름 |
| **위험 요인** | 스택 오버플로우 (`StackOverflowError`) | 무한 루프 발생 가능성 |
| **코드 특징** | 가독성이 높고 점화식 표현에 유용 (트리/그래프) | 상대적으로 복잡해질 수 있으나 자원 효율적 |


```python
# n! 을 재귀함수로 구현
def fact(n):
  # 1. Base case (재귀 호출 종료)
  if n <= 1:
    print(f"factorial({n}) 호출 -> 1반환 (탈출 조건 도착)") # 변환한값 표현
    return 1

  # 2 Recursive Step (재귀 호출)
  print(f"factorial({n}) 호출 -> {n} * factorial({n-1}) 계산 대기중 ....")
  result = n * fact(n-1)
  print(f"factorial({n}) 완료 -> {result}반환")
  return result

# 재귀함수 호출
fact(3)
```

```python
# 하노이탑 문제의 재귀적 합수 구현( 그 원판 옮기기 )
def hanoi_tower(n,fr,temp,to): # n원판개수 fr시작 temp중간지점 to도착지
  # 1. Base case (재귀함수 호출 종료 조건 - 오버플로우가 발생하지 않도록)
  if n == 1:
    print(f"원반 1: %s ---> %s" %(fr,to))

  # 2. recursive case
  else:
    hanoi_tower(n-1,fr,to,temp) # 1단계
    print(f"원반 %d: %s ---> %s" %(n,fr,to)) # 2단계  %d는 숫자 s는 문자
    hanoi_tower(n-1,temp,fr,to) # 3단계

# 재귀함수 호출(A->C로 이동, B는 보조 막대)
hanoi_tower(3, 'A','B','C')
```
